```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

RUNS_DIR = PROJECT_ROOT / "runs"

files = {
    "Gemma-3-4B Base": RUNS_DIR / "gemma-3-4b-it_spider_results.csv",
    "Gemma-3-4B Tuned": RUNS_DIR / "gemma-3-4b-tuned_dev_all_spider_results.csv",
    "Mistral-7B Instruct": RUNS_DIR / "Mistral-7B-Instruct-v0.2_spider_results.csv",
}

# Load data
dfs = {}
for name, path in files.items():
    if path.exists():
        dfs[name] = pd.read_csv(path)
        print(f"Loaded {name}: {len(dfs[name])} rows")
    else:
        print(f"Missing file for {name}: {path}")

# Build summary table
summary_rows = []
for name, df in dfs.items():
    em_pct = df["EM"].mean() * 100 if "EM" in df.columns else None
    ex_pct = df["EX"].mean() * 100 if "EX" in df.columns else None
    avg_latency = df["latency_s"].mean() if "latency_s" in df.columns else None

    summary_rows.append({
        "Model": name,
        "EM %": em_pct,
        "EX %": ex_pct,
        "Avg Latency (s)": avg_latency,
        "Rows": len(df)
    })

summary_df = pd.DataFrame(summary_rows)
display(summary_df)

plt.figure(figsize=(8, 5))
plt.bar(summary_df["Model"], summary_df["EM %"])
plt.ylabel("Exact Match (%)")
plt.title("Exact Match Comparison")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(summary_df["Model"], summary_df["EX %"])
plt.ylabel("Execution Accuracy (%)")
plt.title("Execution Accuracy Comparison")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(summary_df["Model"], summary_df["Avg Latency (s)"])
plt.ylabel("Average Latency (seconds)")
plt.title("Average Inference Latency")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()
```
